```python
receiving_stats_ppr = receiving_stats.with_columns(
    (
        pl.col("receptions")
        + (pl.col("receiving_yards") * 0.1)
        + (pl.col("receiving_tds") * 6)
        + (pl.col('rushing_yards') * 0.1)
        + (pl.col('rushing_tds') * 6)
        - (pl.col("receiving_fumbles_lost") * 2)
        - (pl.col("rushing_fumbles_lost") * 2)
    ).alias("calculated_ppr")
)

receiving_stats_ppr
```

```python
receiving_stats_ppr = receiving_stats_ppr.with_columns(
    (pl.col("calculated_ppr") - pl.col("fantasy_points_ppr")).alias("ppr_difference")
)
receiving_stats_ppr.select([
    'player_display_name',
    'week',
    'fantasy_points_ppr',
    'calculated_ppr',
    'ppr_difference'
]).sort('ppr_difference')
```

```python
ppr_differences = receiving_stats_ppr.filter(
    pl.col("ppr_difference") != 0
)

ppr_differences.select([
    'player_display_name',
    'week',
    'receptions',
    'receiving_yards',
    'receiving_tds',
    'receiving_fumbles_lost',
    'receiving_2pt_conversions',
    'rushing_yards',
    'rushing_tds',
    'rushing_fumbles_lost',
    'rushing_2pt_conversions',
    'special_teams_tds',
    'fantasy_points_ppr',
    'calculated_ppr',
    'ppr_difference'
]).sort('ppr_difference')
```

```python
player_averages = receiving_stats_ppr.group_by('player_display_name').agg([
    pl.col("calculated_ppr").mean().round(2).alias("avg_calculated_ppr"),
    pl.col("receptions").mean().round(2).alias("avg_receptions"),
    pl.col("receiving_yards").mean().round(2).alias("avg_receiving_yards"),
]).sort("avg_calculated_ppr", descending=True)

player_averages
```

```python
lamb_rolling = (
    lamb_ppr
    .sort("week")
    .with_columns(
        pl.col("calculated_ppr")
        .rolling_mean(window_size=3)
        .round(2)
        .alias("rolling_3_ppr"),

        pl.col("receptions")
        .rolling_mean(window_size=3)
        .round(2)
        .alias("rolling_3_receptions"),

        pl.col("receiving_yards")
        .rolling_mean(window_size=3)
        .round(2)
        .alias("rolling_3_receiving_yards")
    )
)

lamb_rolling.select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'rolling_3_ppr',
    'receptions',
    'rolling_3_receptions',
    'receiving_yards',
    'rolling_3_receiving_yards'
])
```

```python
player_rolling_ppr = (
    receiving_stats_ppr
    .sort(["player_display_name", "week"])
    .with_columns(
        pl.col("calculated_ppr")
        .rolling_mean(window_size=3, min_samples=3)
        .round(2)
        .over("player_display_name")
        .alias("rolling_3_ppr"),

        pl.col("receptions")
        .rolling_mean(window_size=3, min_samples=3)
        .round(2)
        .over("player_display_name")
        .alias("rolling_3_receptions"),

        pl.col("receiving_yards")
        .rolling_mean(window_size=3, min_samples=3)
        .round(2)
        .over("player_display_name")
        .alias("rolling_3_receiving_yards")
    )
)

player_rolling_ppr.filter(pl.col('week') == 15).select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'rolling_3_ppr',
    'receptions',
    'rolling_3_receptions',
    'receiving_yards',
    'rolling_3_receiving_yards'
]).sort('calculated_ppr', descending=True)
```

```python
player_rolling_ppr.filter(pl.col('player_display_name') == 'Khalil Shakir').select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'rolling_3_ppr',
    'receptions',
    'rolling_3_receptions',
    'receiving_yards',  
    'rolling_3_receiving_yards'
]).sort('week')
```

```python
player_prediction_features = (
    receiving_stats_ppr
    .sort(["player_display_name", "week"])
    .with_columns(
        pl.col("calculated_ppr")
        .shift(1)
        .rolling_mean(window_size=3, min_samples=3)
        .round(2)
        .over("player_display_name")
        .alias("prev_rolling_3_ppr"),

        pl.col("targets")
        .shift(1)
        .rolling_mean(window_size=3, min_samples=3)
        .round(2)
        .over("player_display_name")
        .alias("prev_rolling_3_targets"),

        pl.col("receptions")
        .shift(1)
        .rolling_mean(window_size=3, min_samples=3)
        .round(2)
        .over("player_display_name")
        .alias("prev_rolling_3_receptions"),

        pl.col("receiving_yards")
        .shift(1)
        .rolling_mean(window_size=3, min_samples=3)
        .round(2)
        .over("player_display_name")
        .alias("prev_rolling_3_receiving_yards")
    )
)

player_prediction_features.filter(pl.col('player_display_name') == 'Ja\'Marr Chase').select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'prev_rolling_3_ppr',
    'targets',
    'prev_rolling_3_targets',
    'receptions',
    'prev_rolling_3_receptions',
    'receiving_yards',  
    'prev_rolling_3_receiving_yards'
]).sort('week')
```

```python
player_prediction_features = (
    player_prediction_features
    .sort(["player_display_name", "week"])
    .with_columns(
        (
            pl.col("calculated_ppr").shift(1).cum_sum().over("player_display_name")
            /
            pl.col("calculated_ppr").shift(1).cum_count().over("player_display_name")
        )
        .round(2)
        .alias("prev_season_avg_ppr"),

        (
            pl.col("targets").shift(1).cum_sum().over("player_display_name")
            /
            pl.col("targets").shift(1).cum_count().over("player_display_name")
        )
        .round(2)
        .alias("prev_season_avg_targets")
    )
)

player_prediction_features.filter(pl.col('player_display_name') == 'Jaxon Smith-Njigba').select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'prev_season_avg_ppr',
    'prev_season_avg_targets'
]).sort('week')
```

```python
baseline_predictions = (
    player_prediction_features
    .with_columns(
        pl.col("prev_rolling_3_ppr").alias("baseline_prediction"),
        pl.col("prev_season_avg_ppr").alias("baseline_prediction_season")
    )
)

baseline_predictions.filter(pl.col('player_display_name') == 'Jaxon Smith-Njigba').select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'baseline_prediction',
    'baseline_prediction_season',
    'prev_rolling_3_ppr',
    'prev_rolling_3_receptions',
    'prev_rolling_3_receiving_yards',
    'prev_season_avg_targets'
])
```

```python
baseline_prediction_clean = baseline_predictions.filter(
    (pl.col("prev_rolling_3_ppr").is_not_null()) & (pl.col("week") <= 18)
)

baseline_prediction_clean = baseline_prediction_clean.with_columns(
    (pl.col("baseline_prediction") - pl.col("calculated_ppr"))
    .abs()
    .round(2)
    .alias("absolute_error"),

    (pl.col("calculated_ppr") - pl.col("baseline_prediction"))
    .round(2)
    .alias("signed_error"),

    (pl.col("calculated_ppr") - pl.col("prev_season_avg_ppr"))
    .abs()
    .round(2)
    .alias("season_avg_error"),

    (pl.col("calculated_ppr") - pl.col("prev_season_avg_ppr"))
    .round(2)
    .alias("signed_season_avg_error")
)

baseline_prediction_clean.filter(pl.col('player_display_name') == 'Justin Jefferson').select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'baseline_prediction',
    'baseline_prediction_season',
    'signed_error',
    'absolute_error',
    'season_avg_error',
    'signed_season_avg_error'
])
```

```python
mean_abs_error = baseline_prediction_clean.filter((pl.col('week') >= 15) & (pl.col('week') <= 18)).select(
    pl.col("absolute_error").mean().round(2).alias("mean_absolute_error")
)
mean_abs_error
```

Across all evaluated wide receiver weeks, the previous-3-game rolling average baseline had a mean absolute error of 4.58 PPR points. This means that, on average, this baseline prediction was off by 4.58 fantasy points from the receiver’s actual calculated PPR score.

```python
mean_season_avg_error = baseline_prediction_clean.filter((pl.col('week') >= 15) & (pl.col('week') <= 18)).select(
    pl.col("season_avg_error").mean().round(2).alias("mean_season_avg_error")
)
mean_season_avg_error
```

Across all evaluated wide receiver weeks, the season-to-date baseline had an mean absolute error of 4.4 PPR points. THis means that, on average, this baseline prediction was off by 4.4 fantasy points from the receiver's actual calculated PPR score.

```python
dataset = (
    baseline_prediction_clean.select([
        'player_display_name',
        'season',
        'week',
        'team',
        'opponent_team',
        'prev_rolling_3_ppr',
        'prev_rolling_3_targets',
        'prev_rolling_3_receptions',
        'prev_rolling_3_receiving_yards',
        'prev_season_avg_ppr',
        'prev_season_avg_targets',
        'calculated_ppr',
    ])
)

dataset = dataset.filter(
    (pl.col("prev_rolling_3_ppr").is_not_null()) & 
    (pl.col("prev_rolling_3_targets").is_not_null()) & 
    (pl.col("prev_rolling_3_receptions").is_not_null()) & 
    (pl.col("prev_rolling_3_receiving_yards").is_not_null()) & 
    (pl.col("prev_season_avg_ppr").is_not_null()) & 
    (pl.col("prev_season_avg_targets").is_not_null()) &
    (pl.col("calculated_ppr").is_not_null())
)

dataset
```

```python
print(dataset.shape)
print(dataset.head(10))
print(dataset.columns)
```

```python
import sklearn
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error


X_train = dataset.filter((pl.col('week') >= 4) & (pl.col('week') <= 14)).select([
    'prev_rolling_3_ppr',
    'prev_rolling_3_targets',
    'prev_rolling_3_receptions',
    'prev_rolling_3_receiving_yards',
    'prev_season_avg_ppr',
    'prev_season_avg_targets'
]).to_numpy()

X_test = dataset.filter((pl.col('week') >= 15) & (pl.col('week') <= 18)).select([
    'prev_rolling_3_ppr',
    'prev_rolling_3_targets',
    'prev_rolling_3_receptions',
    'prev_rolling_3_receiving_yards',
    'prev_season_avg_ppr',
    'prev_season_avg_targets'
]).to_numpy()

y_train = dataset.filter((pl.col('week') >= 4) & (pl.col('week') <= 14)).select('calculated_ppr').to_numpy().ravel()
y_test = dataset.filter((pl.col('week') >= 15) & (pl.col('week') <= 18)).select('calculated_ppr').to_numpy().ravel()

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
print(f"Mean Absolute Error: {mae}")
```

```python
test_results = (
    dataset
    .filter((pl.col('week') >= 15) & (pl.col('week') <= 18))
    .with_columns(
        pl.Series("linear_regression_prediction", y_pred)
    )
)

test_results = test_results.with_columns(
    (pl.col("linear_regression_prediction") - pl.col("calculated_ppr"))
    .abs()
    .round(2)
    .alias("linear_regression_absolute_error"),

    (pl.col("calculated_ppr") - pl.col("linear_regression_prediction"))
    .round(2)
    .alias("linear_regression_signed_error")
)

test_results.select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'linear_regression_prediction',
    'linear_regression_absolute_error',
    'linear_regression_signed_error',
    "prev_rolling_3_ppr",
    "prev_season_avg_ppr",
    "prev_season_avg_targets",
]).sort('linear_regression_absolute_error', descending=True).head(20)
```

```python
test_results = test_results.with_columns(
    (pl.col("prev_rolling_3_ppr") - pl.col("calculated_ppr"))
    .abs()
    .round(2)
    .alias("rolling_baseline_absolute_error"),

    (pl.col("prev_season_avg_ppr") - pl.col("calculated_ppr"))
    .abs()
    .round(2)
    .alias("season_baseline_absolute_error")
)

test_results.select([
    "player_display_name",
    "week",
    "calculated_ppr",
    "linear_regression_prediction",
    "linear_regression_absolute_error",
    "rolling_baseline_absolute_error",
    "season_baseline_absolute_error",
]).sort("linear_regression_absolute_error", descending=True).head(20)
```

I think the linear regression model also suffered from random spikes in performances. From the results, we see that players significantly performed better than the linear model prediction, especially in weeks where the player had a random spike in points. More often then not, the linear regression absolute error was higher than the rolling baseline absolute error and the season baseline absolute error

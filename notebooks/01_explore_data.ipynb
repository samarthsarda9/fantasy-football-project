```python
import pandas as pd
import nflreadpy as nfl
import polars as pl
```

```python
weekly = nfl.load_player_stats([2025])
weekly
```

```python
print(type(weekly))
print(weekly.shape)
print(weekly.columns)
```

```python
receivers = weekly.filter(weekly['position'] == 'WR')
receivers
```

```python
receiving_stats = receivers.select([
    "player_display_name",
    "season",
    "week",
    "team",
    "opponent_team",
    "receptions",
    "targets",
    "receiving_yards",
    "receiving_tds",
    "receiving_fumbles_lost",
    "receiving_2pt_conversions",
    "special_teams_tds",
    "rushing_yards",
    "rushing_tds",
    "rushing_fumbles_lost",
    "rushing_2pt_conversions",
    "fantasy_points_ppr",
])

receiving_stats
```

```python
lamb = receiving_stats.filter((pl.col('player_display_name') == 'CeeDee Lamb'))
lamb
```

```python
lamb.sort("targets", descending=True)
```

```python
lamb.sort("fantasy_points_ppr", descending=True)
```

```python
lamb_week_1 = lamb.filter((pl.col('week') == 1))
lamb_week_1
```

```python
receiving_stats_ppr = receiving_stats.with_columns(
    (
        pl.col("receptions")
        + (pl.col("receiving_yards") * 0.1)
        + (pl.col("receiving_tds") * 6)
        + (pl.col('rushing_yards') * 0.1)
        + (pl.col('rushing_tds') * 6)
        - (pl.col("receiving_fumbles_lost") * 2)
        - (pl.col("rushing_fumbles_lost") * 2)
    ).alias("calculated_ppr")
)

receiving_stats_ppr
```

```python
lamb_ppr = receiving_stats_ppr.filter((pl.col('player_display_name') == 'CeeDee Lamb'))
lamb_ppr.select([
    'player_display_name',
    'week',
    'receptions',
    'receiving_yards',
    'receiving_tds',
    'receiving_fumbles_lost',
    'fantasy_points_ppr',
    'calculated_ppr'
])
# lamb_ppr.sort('week')
# lamb_ppr
```

```python
jsn_ppr = receiving_stats_ppr.filter(pl.col('player_display_name') == 'Jaxon Smith-Njigba')
jsn_summary = jsn_ppr.select([
    'player_display_name',
    'week',
    'receptions',
    'receiving_yards',
    'receiving_tds',
    'receiving_fumbles_lost',
    'rushing_yards',
    'rushing_tds',
    'rushing_fumbles_lost',
    'fantasy_points_ppr',
    'calculated_ppr'
])
pl.Config.set_tbl_rows(50)
jsn_summary
```

```python
puka = receiving_stats_ppr.filter(pl.col('player_display_name') == 'Puka Nacua')
puka_summary = puka.select([
    'player_display_name',
    'week',
    'receptions',
    'receiving_yards',
    'receiving_tds',
    'receiving_fumbles_lost',
    'rushing_yards',
    'rushing_tds',
    'rushing_fumbles_lost',
    'fantasy_points_ppr',
    'calculated_ppr'
])
pl.Config.set_tbl_rows(50)
puka_summary
```

```python
receiving_stats_ppr = receiving_stats_ppr.with_columns(
    (pl.col("calculated_ppr") - pl.col("fantasy_points_ppr")).alias("ppr_difference")
)
receiving_stats_ppr.select([
    'player_display_name',
    'week',
    'fantasy_points_ppr',
    'calculated_ppr',
    'ppr_difference'
]).sort('ppr_difference')
```

```python
ppr_differences = receiving_stats_ppr.filter(
    pl.col("ppr_difference") != 0
)

ppr_differences.select([
    'player_display_name',
    'week',
    'receptions',
    'receiving_yards',
    'receiving_tds',
    'receiving_fumbles_lost',
    'receiving_2pt_conversions',
    'rushing_yards',
    'rushing_tds',
    'rushing_fumbles_lost',
    'rushing_2pt_conversions',
    'special_teams_tds',
    'fantasy_points_ppr',
    'calculated_ppr',
    'ppr_difference'
]).sort('ppr_difference')
```

```python
player_averages = receiving_stats_ppr.group_by('player_display_name').agg([
    pl.col("calculated_ppr").mean().round(2).alias("avg_calculated_ppr"),
    pl.col("receptions").mean().round(2).alias("avg_receptions"),
    pl.col("receiving_yards").mean().round(2).alias("avg_receiving_yards"),
]).sort("avg_calculated_ppr", descending=True)

player_averages
```

```python
lamb_rolling = (
    lamb_ppr
    .sort("week")
    .with_columns(
        pl.col("calculated_ppr")
        .rolling_mean(window_size=3)
        .round(2)
        .alias("rolling_3_ppr"),

        pl.col("receptions")
        .rolling_mean(window_size=3)
        .round(2)
        .alias("rolling_3_receptions"),

        pl.col("receiving_yards")
        .rolling_mean(window_size=3)
        .round(2)
        .alias("rolling_3_receiving_yards")
    )
)

lamb_rolling.select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'rolling_3_ppr',
    'receptions',
    'rolling_3_receptions',
    'receiving_yards',
    'rolling_3_receiving_yards'
])
```

```python
player_rolling_ppr = (
    receiving_stats_ppr
    .sort("player_display_name")
    .with_columns(
        pl.col("calculated_ppr")
        .rolling_mean(window_size=3)
        .round(2)
        .over("player_display_name")
        .alias("rolling_3_ppr"),

        pl.col("receptions")
        .rolling_mean(window_size=3)
        .round(2)
        .over("player_display_name")
        .alias("rolling_3_receptions"),

        pl.col("receiving_yards")
        .rolling_mean(window_size=3)
        .round(2)
        .over("player_display_name")
        .alias("rolling_3_receiving_yards")
    )
)

player_rolling_ppr.filter(pl.col('week') == 15).select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'rolling_3_ppr',
    'receptions',
    'rolling_3_receptions',
    'receiving_yards',
    'rolling_3_receiving_yards'
]).sort('calculated_ppr', descending=True)
```

```python
player_rolling_ppr.filter(pl.col('player_display_name') == 'Khalil Shakir').select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'rolling_3_ppr',
    'receptions',
    'rolling_3_receptions',
    'receiving_yards',  
    'rolling_3_receiving_yards'
]).sort('week')
```

```python
player_prediction_features = (
    receiving_stats_ppr
    .sort(["player_display_name", "week"])
    .with_columns(
        pl.col("calculated_ppr")
        .shift(1)
        .rolling_mean(window_size=3)
        .round(2)
        .over("player_display_name")
        .alias("prev_rolling_3_ppr"),

        pl.col("receptions")
        .shift(1)
        .rolling_mean(window_size=3)
        .round(2)
        .over("player_display_name")
        .alias("prev_rolling_3_receptions"),

        pl.col("receiving_yards")
        .shift(1)
        .rolling_mean(window_size=3)
        .round(2)
        .over("player_display_name")
        .alias("prev_rolling_3_receiving_yards")
    )
)

player_prediction_features.filter(pl.col('player_display_name') == 'Ja\'Marr Chase').select([
    'player_display_name',
    'week',
    'calculated_ppr',
    'prev_rolling_3_ppr',
    'receptions',
    'prev_rolling_3_receptions',
    'receiving_yards',  
    'prev_rolling_3_receiving_yards'
]).sort('week')
```
